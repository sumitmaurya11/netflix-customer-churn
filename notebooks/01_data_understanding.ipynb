```python
#Preparing the Dataset for Machine Learning

df_clean.columns.tolist()
```

```python
#customer_id is just an identifier. It doesn't help predict whether someone will churn.

model_df = df_clean.drop(
    columns=["customer_id"]
).copy()
```

```python
#Converting engineered categorical features to strings for easier one-hot encoding later on

categorical_features = [
    "gender",
    "subscription_type",
    "region",
    "device",
    "payment_method",
    "favorite_genre",
    "inactivity_level",
    "watch_time_level",
    "fee_level"
]

for col in categorical_features:
    model_df[col] = model_df[col].astype(str)
```

```python
#spliting the dataset into features and target variable

X = model_df.drop(columns=["churned"])
y = model_df["churned"]
```

```python
print("Features:", X.shape)
print("Target:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())
```

```python
#Train/Test Split

from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
print("Training churn rate:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting churn rate:")
print(y_test.value_counts(normalize=True).round(3))
```

```python
#Building the Preprocessing Pipeline

#numerical and categorical features


numeric_features = [
    "age",
    "monthly_fee",
    "watch_hours",
    "last_login_days",
    "number_of_profiles",
    "avg_watch_time_per_day",
    "engagement_score"
]

categorical_features = [
    "gender",
    "subscription_type",
    "region",
    "device",
    "payment_method",
    "favorite_genre",
    "inactivity_level",
    "watch_time_level",
    "fee_level"
]
```

```python
#preprocessing tools

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
#StandardScaler puts numerical variables on a similar scale.


numeric_transformer = StandardScaler()
```

```python
#categorical preprocessing


categorical_transformer = OneHotEncoder(
    handle_unknown="ignore"
)
```

```python
#Combining everything

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
```

```python
logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            LogisticRegression(
                max_iter=1000,
                random_state=42
            )
        )
    ]
)
```

```python
#Train the model

logistic_model.fit(
    X_train,
    y_train
)
```

```python
#predictions

y_pred = logistic_model.predict(X_test)
```

```python
#churn probabilities

y_proba = logistic_model.predict_proba(X_test)[:, 1]
```

```python
#evaluation metrics

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_proba)

print(f"Accuracy : {accuracy:.3f}")
print(f"Precision: {precision:.3f}")
print(f"Recall   : {recall:.3f}")
print(f"F1 Score : {f1:.3f}")
print(f"ROC-AUC  : {roc_auc:.3f}")
```

```python
#classification report

print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Stayed", "Churned"]
    )
)
```

```python
#Confusion matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Stayed", "Churned"],
    yticklabels=["Stayed", "Churned"]
)

plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
#Cross-Validation

from sklearn.model_selection import StratifiedKFold, cross_validate
```

```python
#Create 5-fold stratified CV


cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}
```

```python
models = {
    "Logistic Regression": logistic_model,
    "Decision Tree": decision_tree_model,
    "Random Forest": random_forest_model
}

cv_results = []

for name, model in models.items():

    scores = cross_validate(
        model,
        X,
        y,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    cv_results.append({
        "Model": name,
        "Accuracy": scores["test_accuracy"].mean(),
        "Precision": scores["test_precision"].mean(),
        "Recall": scores["test_recall"].mean(),
        "F1 Score": scores["test_f1"].mean(),
        "ROC-AUC": scores["test_roc_auc"].mean()
    })
```

```python
cv_results_df = pd.DataFrame(cv_results)

cv_results_df.round(3).sort_values(
    "ROC-AUC",
    ascending=False
)
```

```python
#Selecting & Retraining the Final Model

best_model_name = (
    cv_results_df
    .sort_values("ROC-AUC", ascending=False)
    .iloc[0]["Model"]
)

print("Selected model:", best_model_name)
```
